# 1. Variables, Typing & Naming Conventions

In Python, you assign variables without explicit type declarations. The type of the variable is determined dynamically at runtime based on the value it holds. However, Python is **strongly typed**, meaning it won't automatically coerce types in ways that don't make sense (like adding a string to an integer).

## Variable Naming Conventions (PEP 8 Standard)
Python has strict rules and strong community conventions for naming variables:
*   **Use `snake_case`**: All lowercase letters with words separated by underscores (`_`).
*   **Be Descriptive**: `student_age` is much better than `a` or `sa`.
*   **Rules**: Variable names cannot start with a number (e.g., `1st_place` is invalid) and cannot contain spaces.
*   **Avoid Reserved Keywords**: Do not use words like `list`, `str`, `def`, `class`, or `for` as variable names.
*   **Constants**: Variables that should never change are written in `UPPER_CASE_WITH_UNDERSCORES`.


````python
# Good variable names (Pythonic)
student_age = 25
max_speed_limit = 120
is_active = True
PI_VALUE = 3.14159  # Constant convention

# Bad (but valid) variable names
a = 25           # Too vague
StudentAge = 25  # PascalCase is reserved for Class names in Python
studentAge = 25  # camelCase is used in JavaScript/Java, but un-Pythonic

# You can assign multiple variables at once
a, b, c = 1, 2, 3

# You can check the type of a variable using the type() function.
print(f"student_age is {student_age} (Type: {type(student_age)})")

# Reassigning to a different type is allowed (Dynamically typed)
student_age = "Twenty Five"
print(f"student_age is now '{student_age}' (Type: {type(student_age)})")
````

# 3. Built-in Data Structures

Python provides robust data structures. Choosing the right one is key to writing efficient code.

## Lists (Mutable, ordered)
Lists are the workhorse of Python. They can hold mixed data types and can grow or shrink dynamically.


````python
my_list = [10, 20, 30, 40, 50, 60, 70]

# 1. Indexing (getting individual items)
print("First item:", my_list[0])
print("Last item:", my_list[-1])

# 2. Slicing [start:stop:step] (getting subsets)
print("Items from index 1 to 3:", my_list[1:4])
print("Every second item:", my_list[::2])
print("Reversed list:", my_list[::-1])

# 3. Iterating (getting items one by one)
print("\nIterating through the list:")
for item in my_list[:3]:  # Iterating over a slice
    print(f"Item: {item}")

# 4. Modifying
my_list.append(80)       # Add to end
my_list.insert(1, 15)    # Insert 15 at index 1
my_list[0] = 5           # Change the first element
print("\nAfter additions and updates:", my_list)
````

## Tuples (Immutable, ordered)
Tuples are faster than lists and protect data from accidental modification.


````python
my_tuple = (100, 200, 300, 400, 500)

# 1. Indexing and Slicing (same as lists)
print("Second item in tuple:", my_tuple[1])
print("Last three items:", my_tuple[-3:])

# 2. Iterating one by one
print("\nIterating through the tuple:")
for val in my_tuple:
    print(val * 2) # You can use the values, but not change the tuple itself

# 3. Unpacking (Very common in Python)
point = (34.5, -118.2, 100)
lat, lon, alt = point
print(f"\nLatitude: {lat}, Longitude: {lon}, Altitude: {alt}")

# my_tuple[0] = 999  # TypeError: 'tuple' object does not support item assignment
````

## Dictionaries (Key-Value, mutable)
Dictionaries map unique keys to values. They are highly optimized for fast lookups.


````python
student = {
    "name": "Alice",
    "major": "Computer Science",
    "gpa": 3.8
}

# 1. Getting items
print("Student name:", student["name"])
print("Graduation year:", student.get("grad_year", "Not specified")) # Safe get

# 2. Updating values and adding new keys
student["gpa"] = 3.9               # Update existing key
student["grad_year"] = 2025        # Add new key-value pair
student.update({"courses": ["Math", "Physics"], "active": True}) # Update multiple
print("\nUpdated dictionary:", student)

# 3. Removing items
removed_gpa = student.pop("gpa")
print(f"\nRemoved GPA ({removed_gpa}). Dict now:", student)

# 4. Iterating over keys, values, and items
print("\nIterating over keys:")
for key in student.keys():
    print(key)

print("\nIterating over items (key-value pairs):")
for key, value in student.items():
    print(f"{key.capitalize()}: {value}")
````

````python
# EASY: Simple check
temperature = 35
if temperature > 30:
    print("Easy: It's hot outside.")

# MEDIUM: Multiple conditions (if/elif/else)
score = 85
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"
print(f"Medium: With a score of {score}, the grade is {grade}.")

# ADVANCED: Nested conditionals and logical operators (and/or)
user = {"is_logged_in": True, "role": "admin", "age": 25}

if user["is_logged_in"]:
    if user["role"] == "admin" and user["age"] >= 18:
        print("Advanced: Access granted to adult admin dashboard.")
    elif user["role"] == "user":
        print("Advanced: Access granted to standard dashboard.")
    else:
        print("Advanced: Role restricted.")
else:
    print("Advanced: Please log in.")
````

## `for` Loops
Used for iterating over a sequence (list, tuple, dictionary, string, or range).


````python
# EASY: Print items in a list
fruits = ["apple", "banana", "cherry"]
print("Easy for loop:")
for fruit in fruits:
    print(f"- {fruit}")

# MEDIUM: Calculate aggregate values and filtering
numbers = [10, 15, 20, 25, 30]
even_sum = 0
for num in numbers:
    if num % 2 == 0:  # Check if even
        even_sum += num
print(f"\nMedium for loop: Sum of even numbers is {even_sum}")

# ADVANCED: Nested loops (e.g., flattening a matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
flattened = []
for row in matrix:
    for item in row:
        flattened.append(item)
print(f"\nAdvanced for loop: Flattened matrix -> {flattened}")
````

## `while` Loops
Used to repeat a block of code as long as a condition remains true.


````python
# EASY: Simple countdown
count = 3
print("Easy while loop:")
while count > 0:
    print(count)
    count -= 1
print("Go!")

# MEDIUM: Find the first power of 2 greater than 1000
value = 1
while value <= 1000:
    value *= 2
print(f"\nMedium while loop: First power of 2 > 1000 is {value}")

# ADVANCED: Loop with 'break', 'continue', and flag variables
import random
print("\nAdvanced while loop: Simulating a random walk until distance > 3")
position = 0
steps = 0

while True: # Infinite loop, must use break to exit
    step = random.choice([-1, 1])
    position += step
    steps += 1
    
    if position == 0:
        continue # Skip the print if we are back at the start
        
    print(f"Step {steps}, Position: {position}")
    
    if abs(position) >= 3:
        print("Reached target distance. Stopping.")
        break # Exit the loop
````

# 5. Functions (def)

## Why do we need functions?
As your code grows, writing everything top-to-bottom becomes impossible to manage. Functions solve this by offering:
1.  **Reusability (DRY - Don't Repeat Yourself)**: Write the logic once and use it many times.
2.  **Modularity**: Break down complex, giant problems into small, manageable chunks.
3.  **Abstraction**: You don't need to know *how* `print()` works internally to use it. You can create your own abstractions.
4.  **Testability & Debugging**: It is much easier to test and fix a small 5-line function than a 500-line script.

## Function Naming Conventions
Just like variables, function names must follow **PEP 8**:
*   Use `snake_case` (lowercase with underscores).
*   Use **action verbs** because functions *do* things (e.g., `calculate_area()`, `get_user_data()`, `is_valid()`).

## Writing Functions: From Simple to Advanced


### 1. Simple Function (No Arguments)
The `def` keyword defines a function. Code inside is indented.


````python
def greet_world():
    # This code only runs when the function is called
    print("Hello, World!")

# Calling the function
greet_world()
````

### 2. Single Argument, Single Output
Functions take inputs (arguments) and give back outputs (using `return`).
If you `print()` inside a function, it just displays text. If you `return`, you can save the result to a variable.


````python
def square_number(number):
    """Calculates and returns the square of a number."""
    result = number ** 2
    return result

# We pass 5 as an argument, and save the returned output to 'squared_val'
squared_val = square_number(5)
print(f"The square of 5 is: {squared_val}")
````

### 3. Multiple Arguments: Order Matters (Positional vs. Keyword)
When passing multiple arguments, Python needs to know which value belongs to which parameter.

*   **Positional Arguments**: Assigned based on the **order** they are passed.
*   **Keyword Arguments**: Explicitly state the parameter name (order doesn't matter).
*   **Default Arguments**: You can provide a default value so the user doesn't have to provide it.


````python
def create_profile(first_name, last_name, role="User"):
    """Creates a user profile string."""
    return f"{last_name.upper()}, {first_name} - Role: {role}"

# 1. Positional Arguments (Order matters heavily!)
# If we mess up the order, the data is wrong.
print("Positional:", create_profile("Alice", "Smith")) 
print("Positional (Wrong Order!):", create_profile("Smith", "Alice"))

# 2. Keyword Arguments (Order does NOT matter!)
print("Keyword:", create_profile(last_name="Johnson", first_name="Bob"))

# 3. Using the Default Argument
# Notice we didn't provide a role, so it defaults to "User"
print("Default Role:", create_profile("Charlie", "Brown"))

# Overriding the Default Argument
print("Overridden Role:", create_profile("Diana", "Prince", role="Admin"))
````

### 4. Functions with Multiple Outputs (Tuples)
Often, you want a function to return more than one piece of data. In Python, you can simply return them separated by commas. Under the hood, Python bundles them into a **Tuple**.


````python
def get_min_and_max(numbers_list):
    """Returns both the smallest and largest number in a list."""
    if not numbers_list:
        return None, None
    
    lowest = min(numbers_list)
    highest = max(numbers_list)
    
    # Returning multiple values
    return lowest, highest

# Calling the function and unpacking the tuple into two variables
data = [15, 2, 99, 43, 8]
min_val, max_val = get_min_and_max(data)

print(f"Data: {data}")
print(f"Minimum: {min_val}")
print(f"Maximum: {max_val}")
````

### 5. Advanced Practical Example
Let's combine concepts: Lists, loops, conditionals, and functions.
We will write a function that cleans up dirty data (e.g., a list of messy text strings) and returns the clean data.


````python
def clean_text_data(raw_data, make_lowercase=True):
    """
    Takes a list of messy strings, strips whitespace, and optionally 
    converts them to lowercase. Skips empty strings.
    """
    clean_list = []
    
    for item in raw_data:
        # Check if the item is a string before trying string methods
        if not isinstance(item, str):
            continue
            
        stripped_item = item.strip()
        
        # Check truthiness to skip empty strings
        if stripped_item: 
            if make_lowercase:
                clean_list.append(stripped_item.lower())
            else:
                clean_list.append(stripped_item)
                
    return clean_list

# Test our advanced function
messy_inputs = ["  Apple ", "   ", "BANANA  ", 42, "  cherry \n"]

print("Raw Data:", messy_inputs)

# Using default setting (lowercase=True)
print("Cleaned (Default):", clean_text_data(messy_inputs))

# Using keyword argument to change behavior
print("Cleaned (Keep Case):", clean_text_data(messy_inputs, make_lowercase=False))
````

# 6. Object-Oriented Programming (OOP)

## Why do we need OOP?
As programs become highly complex, simply passing lists and dictionaries into functions isn't enough. We need a way to **model the real world** directly in our code. 

**Advantages of OOP:**
1.  **Encapsulation**: It groups related data (variables) and behavior (functions) into a single secure package. You don't have a `dog_name` variable floating around disconnected from a `dog_bark()` function.
2.  **State Management**: It allows you to create independent entities that track their own state over time, without interfering with one another (e.g., User A's shopping cart vs. User B's).
3.  **Inheritance & Polymorphism**: (Advanced) You can build upon existing classes to create new ones without rewriting code.

## The Core Components of OOP
*   **Class (The Blueprint)**: A template that defines what properties and actions a certain type of thing should have. (e.g., A `Car` blueprint).
*   **Object (The Instance)**: An actual, physical creation built from the blueprint. (e.g., *My* specific red Honda Civic).
*   **Attributes (Data)**: Variables that belong to an object. (e.g., `color`, `speed`).
*   **Methods (Behavior)**: Functions that belong to an object. They usually modify the object's attributes or perform an action using them. (e.g., `accelerate()`, `brake()`).

## Naming Conventions
*   **Classes**: Must use **PascalCase** (e.g., `BankAccount`, `DataProcessor`, `Robot`).
*   **Methods and Attributes**: Use standard `snake_case` just like regular variables and functions.

## The Magic `__init__` Method
`__init__` (short for initialize) is a special method called the **constructor**. 
*   **Why it matters**: It is automatically executed the exact moment you create a new object. Its sole purpose is to set up the *initial state* (starting attributes) of that specific object.
*   **The `self` Keyword**: Every method inside a class must have `self` as its very first parameter. `self` refers to the *specific instance* of the object you are currently working with. Without `self`, the object wouldn't know if it's changing its own data or someone else's data.


### Example 1: Basic Class with a Single Method
Let's define a simple blueprint for a `Dog`.


````python
class Dog:
    # 1. The __init__ constructor sets up the initial data
    def __init__(self, name, breed):
        # We attach the arguments to the specific instance using 'self'
        self.name = name       # Attribute
        self.breed = breed     # Attribute

    # 2. A single method that uses the instance's data
    def bark(self):
        return f"{self.name} says Woof!"

# 3. Creating Objects (Instances)
# When we do this, __init__ is called automatically.
dog_1 = Dog("Buddy", "Golden Retriever")
dog_2 = Dog("Lucy", "Poodle")

# 4. Calling Methods
print(dog_1.bark())
print(dog_2.bark())

# Accessing attributes directly
print(f"Dog 2 is a {dog_2.breed}")
````

### Example 2: Advanced Class with Multiple Methods and State Management
Let's build something more practical: A `ShoppingCart` class that maintains state (memory) over time.

**Avoiding Memory Confusion:** By creating a new instance (`cart = ShoppingCart()`), we ensure that `cart` has its own isolated memory space. If we make `cart2 = ShoppingCart()`, it will have a completely separate list of items.


````python
class ShoppingCart:
    def __init__(self, owner_name):
        self.owner = owner_name
        # We initialize an empty list to hold the cart items. 
        # Every new ShoppingCart gets its own fresh list!
        self.items = []  
        self.is_checked_out = False

    def add_item(self, item_name, price):
        """Adds an item to the cart if it hasn't been checked out."""
        if self.is_checked_out:
            print(f"Sorry {self.owner}, you already checked out!")
            return
        
        # Adding data as a dictionary
        self.items.append({"name": item_name, "price": price})
        print(f"Added {item_name} to {self.owner}'s cart.")

    def calculate_total(self):
        """Calculates the sum of all item prices."""
        total = 0
        for item in self.items:
            total += item["price"]
        return total

    def checkout(self):
        """Finalizes the purchase."""
        if not self.items:
            print(f"{self.owner}'s cart is empty. Nothing to buy.")
            return

        total = self.calculate_total()
        self.is_checked_out = True
        print(f"\n--- Checkout for {self.owner} ---")
        print(f"Items bought: {len(self.items)}")
        print(f"Total charged: ${total:.2f}")

    def reset_cart(self):
        """Resets the state to avoid memory confusion if reusing the object."""
        self.items = []
        self.is_checked_out = False
        print(f"{self.owner}'s cart has been emptied and reset.")


# --- Practical Usage ---

# 1. Instantiate separate carts (Memory Isolation)
alice_cart = ShoppingCart("Alice")
bob_cart = ShoppingCart("Bob")

# 2. Add items to Alice's cart
alice_cart.add_item("Laptop", 999.99)
alice_cart.add_item("Mouse", 49.50)

# 3. Add items to Bob's cart (Notice it does not affect Alice's cart)
bob_cart.add_item("Headphones", 89.00)

# 4. Process checkouts
alice_cart.checkout()

# 5. Try to add an item after checkout (Testing state protection)
alice_cart.add_item("Keyboard", 100.00)

# 6. Resetting the cart
alice_cart.reset_cart()
alice_cart.add_item("Keyboard", 100.00)
````

# 7. Error & Exception Handling

## Why is Error Handling Important?
In the real world, data is messy, networks fail, and users enter unexpected inputs. If your program assumes everything will work perfectly, a single bad piece of data will cause a **fatal crash**, stopping the entire script dead in its tracks. 

Error handling (using `try` and `except`) allows your program to:
1.  **Survive Bad Data**: Catch the error, log it, and move on to the next task instead of crashing.
2.  **Provide Useful Feedback**: Tell the user exactly *what* went wrong, instead of spitting out a terrifying wall of red traceback code.
3.  **Execute Cleanup Tasks**: Ensure files or database connections are closed safely, even if an explosion happens in the middle of processing.


### The Problem: A Program That Crashes
Imagine we are writing a numerical computing function to calculate the average temperature from a sensor. 
If the sensor malfunctions and gives us bad data (like a string instead of a number, or an empty list), our naive program will crash.


````python
# NAIVE APPROACH (Will Crash!)
def naive_calculate_average(data_list):
    total = sum(data_list)
    avg = total / len(data_list)
    return avg

# Test 1: Good Data (Works fine)
print("Naive Test 1:", naive_calculate_average([20, 22, 24]))

# Test 2: Bad Data (Sensor returned an empty list)
# Uncomment the line below to see the crash!
# print("Naive Test 2:", naive_calculate_average([])) 
# -> ZeroDivisionError: division by zero

# Test 3: Bad Data (Sensor returned a string)
# Uncomment the line below to see the crash!
# print("Naive Test 3:", naive_calculate_average([20, "ERROR", 24])) 
# -> TypeError: unsupported operand type(s) for +: 'int' and 'str'

print("Notice: If a crash happened above, this line would NEVER print.")
````

### The Solution: Using `try`, `except`, `else`, and `finally`
Let's rebuild that function to be bulletproof.

**How it works:**
*   `try`: Put the code that *might* blow up inside here. Python will attempt to run it.
*   `except`: If an error happens in the `try` block, execution immediately jumps here. You can specify *which* error to catch.
*   `else`: (Optional) Runs *only* if the `try` block succeeded with absolutely no errors.
*   `finally`: (Optional) Runs *every single time*, regardless of success or failure. It is used for mandatory cleanup.


````python
def robust_calculate_average(data_list):
    try:
        # Code that might fail
        total = sum(data_list)
        avg = total / len(data_list)
        
    except ZeroDivisionError:
        # Specifically catching division by zero (empty list)
        print("Error: The sensor provided an empty list. Cannot calculate average.")
        return 0.0
        
    except TypeError as e:
        # Specifically catching bad data types (like a string).
        # We also capture the exact error message into variable 'e'
        print(f"Error: Non-numeric data found in the list. Details: {e}")
        return None
        
    except Exception as e:
        # The generic 'Exception' catches ANYTHING else we didn't think of.
        # It's usually best practice to put this at the very bottom.
        print(f"An unexpected error occurred: {e}")
        return None
        
    else:
        # Runs ONLY if the try block succeeds perfectly
        print("Calculation successful.")
        return avg
        
    finally:
        # Runs EVERY time, pass or fail.
        print("--- Sensor reading cycle completed ---\n")

# Let's run our tests again! This time, NO crashes!

print("Test 1 (Good Data):")
result1 = robust_calculate_average([20, 22, 24])
print(f"Result: {result1}")

print("Test 2 (Empty List):")
result2 = robust_calculate_average([])
print(f"Result: {result2}")

print("Test 3 (String Data):")
result3 = robust_calculate_average([20, "ERROR", 24])
print(f"Result: {result3}")

print("Because we handled the errors, the program survived and reached the end!")
````

### Advanced: Raising Your Own Errors (`raise`)
Sometimes, Python thinks the code is perfectly fine, but *your business logic* dictates it's an error. You can forcefully trigger an error using the `raise` keyword.

For example, a user trying to deposit negative money into a bank account. Python can do the math (`100 + -50 = 50`), but logically, that's invalid.


````python
def deposit_money(balance, amount):
    try:
        if amount < 0:
            # We explicitly trigger a ValueError with a custom message
            raise ValueError("Deposit amount cannot be negative!")
            
        new_balance = balance + amount
        print(f"Deposited ${amount}. New balance: ${new_balance}")
        return new_balance
        
    except ValueError as e:
        # We catch the exact error we just raised
        print(f"Transaction Failed: {e}")
        return balance

# Test a good deposit
deposit_money(100, 50)

# Test a bad deposit (Python math would allow this, but our 'raise' stops it)
deposit_money(100, -20)
````

# 8. Writing "Pythonic" Code

Bridge the gap between "writing code that works" and "writing beautiful Python." Writing Pythonic code means using the language's unique features to make code more readable, concise, and efficient.


### Example 1: `enumerate()`
Instead of manually tracking an index using `range(len())`, `enumerate` gives you both the index and the item simultaneously.


````python
# ❌ Un-Pythonic (The "C/Java" way)
colors = ["Red", "Green", "Blue"]
for i in range(len(colors)):
    print(f"Index {i} is {colors[i]}")

print("-" * 15)

# ✅ Pythonic
for index, color in enumerate(colors):
    print(f"Index {index} is {color}")
````

### Example 2: `zip()`
When you need to iterate through two or more lists at the exact same time, use `zip()`.


````python
# ❌ Un-Pythonic
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

for i in range(len(names)):
    print(f"{names[i]} got a {scores[i]}")

print("-" * 15)

# ✅ Pythonic
for name, score in zip(names, scores):
    print(f"{name} got a {score}")
````

### Example 3: Multiple Variable Assignment & Unpacking
Python allows you to assign multiple variables on a single line, and easily swap them.


````python
# ❌ Un-Pythonic swapping (Requires a temporary variable)
a = 10
b = 20
temp = a
a = b
b = temp

# ✅ Pythonic unpacking (Swap instantly)
x = 10
y = 20
x, y = y, x 
print(f"Swapped: x={x}, y={y}")

# ✅ Pythonic packing/unpacking
data = ["Alice", 25, "Engineer"]
name, age, profession = data
print(f"Unpacked Name: {name}, Age: {age}")
````

### Example 4: List Comprehensions
List comprehensions are a powerful, single-line way to create a new list by applying an operation or a filter to an existing list. They replace clunky `for` loops that just append to empty lists.


````python
# ❌ Un-Pythonic: Appending in a loop
numbers = [1, 2, 3, 4, 5, 6]
evens_squared = []
for num in numbers:
    if num % 2 == 0:
        evens_squared.append(num ** 2)

# ✅ Pythonic: List Comprehension
# Syntax: [expression for item in iterable if condition]
pythonic_evens_squared = [num ** 2 for num in numbers if num % 2 == 0]

print("Pythonic Evens Squared:", pythonic_evens_squared)
````

### Example 5: Dictionary Comprehensions
Similar to list comprehensions, you can create dictionaries on the fly.


````python
# Let's say we want a dictionary where the key is the word, and the value is its length.
words = ["apple", "banana", "cherry", "date"]

# ✅ Pythonic Dictionary Comprehension
# Syntax: {key_expression: value_expression for item in iterable}
word_lengths = {word: len(word) for word in words}

print("Dictionary Comprehension Result:")
for k, v in word_lengths.items():
    print(f"  {k}: {v}")
````

# 9. Library and Dependency Management

Python's real power comes from its massive ecosystem of third-party libraries. You don't have to reinvent the wheel—someone else has likely already solved your problem!

## 1. The Open Source Advantage
Python is Open Source, which means its source code is freely available. 
*   **Massive Ecosystem**: There are over 400,000 packages available on the Python Package Index (PyPI).
*   **Community Driven**: If there's a bug in a popular library like `pandas` or `numpy`, thousands of developers are looking at it, and fixes are pushed rapidly.
*   **Cost**: It's completely free to use for personal and commercial projects.

## 2. What is Package Management?
A package manager is a tool that automates the process of installing, upgrading, configuring, and removing libraries.
*   **`pip`**: The default package installer for Python (stands for "Pip Installs Packages"). It fetches libraries from PyPI.
*   **`conda`**: Another popular package manager (and environment manager), heavily used in Data Science because it handles complex, non-Python dependencies (like C++ libraries for machine learning) very well.

*(Note: The following commands are typically run in your computer's terminal/command prompt, not directly in a python script, though Jupyter notebooks allow you to run terminal commands using the `!` prefix.)*

## 3. How to Install and Check Libraries


````python
# Checking if a library is installed and its version using Python code
try:
    import requests
    print(f"The 'requests' library is installed! Version: {requests.__version__}")
except ImportError:
    print("The 'requests' library is NOT installed.")

try:
    import numpy
    print(f"The 'numpy' library is installed! Version: {numpy.__version__}")
except ImportError:
    print("The 'numpy' library is NOT installed.")

try:
    import inpainting
    print(f"The 'inpainting' library is installed! Version: {inpainting.__version__}")
except ImportError:
    print("The 'inpainting' library is NOT installed.")
````

**Terminal Commands for Package Management:**

*   **Install a new library:**
    `pip install requests`
*   **Install a specific version:**
    `pip install requests==2.28.1`
*   **Check installed version (Terminal):**
    `pip show requests`
*   **List ALL installed libraries:**
    `pip list`
    
## 4. The Golden Rule: Avoid System-Wide Installations
If you run `pip install pandas` on your main computer, it installs it globally. 

**Why is this bad? (Dependency Hell)**
*   Project A needs `pandas` version 1.0.
*   Project B needs `pandas` version 2.0.
*   If you install globally, upgrading `pandas` for Project B will completely break Project A!

## 5. The Solution: Virtual Environments
A virtual environment is an isolated, self-contained directory that holds a specific Python version and specific library versions for ONE project.

*   **`venv` (Built-in Python tool)**: Great for standard Python projects.
*   **`conda` (Anaconda/Miniconda)**: Excellent for data science, handles complex dependencies, and allows you to install completely different *versions of Python itself* easily.

## 6. How to Use Conda Environments (Step-by-Step)
*(Run these in your terminal)*

1.  **Create a new environment** named `my_project` running Python 3.10:
    `conda create --name my_project python=3.10`
2.  **Activate the environment** (This isolates your terminal session):
    `conda activate my_project`
3.  **Install safely**:
    `conda install pandas` (or `pip install pandas`)
4.  **Deactivate** (Go back to the main system):
    `conda deactivate`

## 7. What is a `requirements.txt` file?
A `requirements.txt` file is a plain text file that lists all the libraries (and their exact versions) needed to run your project. This makes your code **reproducible**. If you send your code to a coworker, they need to know what libraries to install!

**How to create one (saving your current environment):**
`pip freeze > requirements.txt`

*(Example output inside the file)*:
```text
numpy==1.24.3
pandas==2.0.1
requests==2.28.1
```

**How to install from one (setting up a new project):**
Your coworker downloads your code and runs:
`pip install -r requirements.txt`

## 8. Summary of Best Practices
1.  **Never** use system Python for projects.
2.  **Always** create a Conda or Venv environment for every new project.
3.  **Always** generate a `requirements.txt` (or Conda `environment.yml`) file so others can reproduce your work without dependency errors.
